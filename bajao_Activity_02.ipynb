```python
path = kagglehub.dataset_download("danielcorredera/student-mat-csv")
print("Path to dataset files: ", path)
```

```python
print(os.listdir(path))
```

```python
file_path = os.path.join(path, "student-mat.csv")
df = pd.read_csv(file_path, sep=';')
```
